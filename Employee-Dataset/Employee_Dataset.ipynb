```python
import numpy as np
import pandas as pd

import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv(f'Messy_Employee_dataset.csv')
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
```

```python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
df.isna().sum()
```

```python
df.head()
```

```python
df['Phone'].astype(str).str.startswith('-').sum()
```

```python
df['Phone'] = df['Phone'].astype(str).str.lstrip('-')
```

```python
df.head()
```

```python
df.info()
```

```python
df['Join_Date'] = pd.to_datetime(df['Join_Date'])
```

```python
df['Status'] = df['Status'].astype('category')
df['Performance_Score'] = df['Performance_Score'].astype('category')
```

```python
df.head()
```

```python
df.info()
```

```python
df[['Department', 'Region']] = df['Department_Region'].str.split('-', expand=True)
```

```python
df['Region'] = df['Region'].astype('category')
df['Department'] = df['Department'].astype('category')
```

```python
df.drop('Department_Region', axis=1, inplace=True)
```

```python
df.head()
```

```python
df['Age'] = df['Age'].astype(int)
```

```python
df.head()
```

```python
df.info()
```

```python
df.head()
```

```python
from pandas.api.types import CategoricalDtype

order = CategoricalDtype(['Poor', 'Average', 'Good', 'Excellent'], ordered=True)

df['Performance_Score'] = df['Performance_Score'].astype(str).astype(order)
```
